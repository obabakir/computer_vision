```python
import matplotlib.pyplot as plt
import numpy as np
```

#Basic line

```python
x = np.array([1, 2, 4, 5])
y = np.array([10, 20, 30, 40])

plt.plot(x, y)
plt.title("Basic line")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.show()
```

# Line Styles and Color

```python
x = np.array([1, 2, 3, 4 ])
y = np.array([3, 8, 1, 6])

plt.plot(y, linestyle = "dotted", color = "green", marker = "o")
plt.title("Styled line Example")
# plt.xlabel("X Value")
# plt.ylabel("Y Value")
plt.show()
```

# Multiple lines in one Graph with Numpy and Matplotlib

```python
from matplotlib.lines import lineStyles
x = np.array([1, 2, 3, 4 ])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([1, 3, 5, 7])

plt.plot(x, y1, label = "Line 1", color = "red", marker = "s")
plt.plot(x, y2, label = "Line 2", color = "blue", linestyle = "--", marker = "o")
plt.title("Multiple lines Example")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.legend()
# legend bu chap burchakdagi  belgi
plt.show()
```

## Bar chart with Numpy and Matplotlib

```python
x = np.array(["Salary", "Expences", "Charity", "Food" ])
y = np.array([5, 9, 1, 7])

plt.bar(x, y, color = "orange")
plt.title("Bar Chart Example")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.show()
```

# Scatter Plot with Numpy and Matplot

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 6, 3, 7])

plt.scatter(x, y, color = "purple", s = 90)
plt.title("Scatter Plot Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

# Histogram with Numpy and Matplot

```python
data = np.array([3, 4, 5, 6, 8, 8, 8, 9, 10])
plt.hist(data, bins = 5, color = "skyblue", edgecolor = "red")
plt.title("Histogram Example")
plt.xlabel("X value")
plt.ylabel("Y value")
plt.show()
```

```python
x = np.array([1, 2, 3, 4 ])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([3, 6, 9, 12])

# figure  bolmasa auto size holatga keladi
plt.figure(figsize=(8, 4)) # 8 => width, 4 => height

plt.subplot(1, 2, 1) # 1 => row, 2 => column, 1 => index (yani birinchi joylashuvi)
plt.plot(x, y1, color = "red")
plt.title(" Plot 1")

plt.subplot(1, 2, 2)  # 1 => row, 2 => column, 2 => index (yani ikkinchi joylashuvi)
plt.bar(x, y2, color = "blue")
plt.title(" Plot 2")

plt.tight_layout()
plt.show()
```

# Save plot example

```python
x = np.array([1, 2, 3, 4])
y = np.array([4, 8, 6, 10])

plt.plot(x, y, color = "green")
plt.title("Save Figure Example")
plt.savefig("my_plot.png")  # in vs code => Save in Current Folder

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery-nogrid')


# make data
x = [1, 2, 3, 4]
colors = plt.get_cmap('Blues')(np.linspace(0.2, 0.7, len(x)))

# plot
fig, ax = plt.subplots()
ax.pie(x, colors=colors, radius=3, center=(4, 4),
       wedgeprops={"linewidth": 1, "edgecolor": "white"}, frame=True)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```

```python
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import axes3d

plt.style.use('_mpl-gallery')

# Make data
X, Y, Z = axes3d.get_test_data(0.05)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_wireframe(X, Y, Z, rstride=10, cstride=10)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
